```python
%%sql
CREATE TABLE customers (
    customer_id INT PRIMARY KEY,
    customer_name VARCHAR(100),
    email VARCHAR(100),
    city VARCHAR(50),
    state VARCHAR(50),
    signup_date DATE,
    gender VARCHAR(20)
);

```

```python
%%sql
CREATE TABLE products (
    product_id INT PRIMARY KEY,
    product_name VARCHAR(100),
    category VARCHAR(50),
    brand VARCHAR(50),
    price DECIMAL(10,2),
    stock_quantity INT
);
```

```python
%%sql
CREATE TABLE orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    order_date DATE,
    order_status VARCHAR(30),
    shipping_city VARCHAR(50),
    FOREIGN KEY (customer_id) REFERENCES customers(customer_id));
```

```python
%%sql
   CREATE TABLE order_items (
    order_item_id INT PRIMARY KEY,
    order_id INT,
    product_id INT,
    quantity INT,
    unit_price DECIMAL(10,2),
    discount DECIMAL(5,2),
    FOREIGN KEY (order_id) REFERENCES orders(order_id),
    FOREIGN KEY (product_id) REFERENCES products(product_id)
);
```

```python
%%sql
CREATE TABLE payments (
    payment_id INT PRIMARY KEY,
    order_id INT,
    payment_date DATE,
    payment_method VARCHAR(30),
    payment_status VARCHAR(30),
    amount DECIMAL(10,2),
    FOREIGN KEY (order_id) REFERENCES orders(order_id)
);
```

```python
%%sql

##  1. Display all products. Sort products by price from highest to lowest.

select * from products 
order by price desc;
```

```python
%%sql

## Q2 – High-Value Products, Find products priced above ₹10,000. Show product name, category, brand, and price. Sort price descending.

select product_name,category,brand,price
from products
where price >10000
order by price desc;
```

```python
%%sql

##  Q3 – Unique Categories Display all unique product categories

select distinct category
from products;
```

```python
%%sql

## Q4 – Recent Orders Display the 10 most recent orders, including order ID, customer ID, order date, and status

select order_id, customer_id, order_date, order_status 
from orders
order by Order_date desc
limit 10;
```

```python
%%sql

## Q5 – Low-Stock Products Find products with stock quantity below 20. Show name, category, stock, and price

select product_name,category,stock_quantity,price
from products
where stock_quantity <20;
```

```python
%%sql

## Q6 – Category Performance For each category, calculate product count, average price, maximum price, and minimum price. Sort by average price descending

select category,
count(Product_name) as product_count,
avg(price) as avg_price,
max(price) as max_price,
min(price) as min_price
from products
group by category
order by avg(price) desc
```

```python
%%sql

## Q7 – Customer Order Count Show customer ID, name, and total orders. Include only customers who placed more than one order 

SELECT
c.customer_id,
c.customer_name,
COUNT(o.order_id) AS total_orders
FROM customers c
JOIN orders o
ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name
HAVING COUNT(o.order_id) > 1;

```

```python
%%sql

## Q8 – Revenue by Product Calculate product revenue as quantity × unit_price × (1 - discount/100). Show product name, total quantity, and revenue. Sort highest revenue first

SELECT
p.product_name,
SUM(oi.quantity) AS total_quantity,
SUM(oi.quantity * oi.unit_price * (1 - oi.discount / 100.0)) AS revenue
FROM order_items oi
JOIN products p
ON oi.product_id = p.product_id
GROUP BY p.product_id, p.product_name
ORDER BY revenue DESC;
```

```python
%%sql

## Order Status Analysis Calculate delivered, cancelled, and returned order counts and each status percentage of total orders

SELECT
order_status,
COUNT(*) AS order_count,
ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) AS percentage
FROM orders
GROUP BY order_status
HAVING order_status IN ('delivered', 'cancelled', 'returned');
```

```python
%%sql

## Q10 – City-wise Sales For each shipping city, calculate order count, revenue, and average order value

SELECT
o.shipping_city,
COUNT(DISTINCT o.order_id) AS order_count,
ROUND(SUM(oi.quantity * oi.unit_price * (1 - oi.discount / 100.0)), 2) AS revenue,
ROUND(
SUM(oi.quantity * oi.unit_price * (1 - oi.discount / 100.0))
/ COUNT(DISTINCT o.order_id), 2
) AS avg_order_value
FROM orders o
JOIN order_items oi
ON o.order_id = oi.order_id
GROUP BY o.shipping_city
ORDER BY revenue DESC;
```

```python
%%sql

 ## Q11 – Top 5 Customers Identify the top 5 customers by revenue. Show customer name, order count, and revenue

SELECT
c.customer_name,
COUNT(DISTINCT o.order_id) AS order_count,
ROUND(SUM(oi.quantity * oi.unit_price * (1 - oi.discount / 100.0)), 2) AS revenue
FROM customers c
JOIN orders o
ON c.customer_id = o.customer_id
JOIN order_items oi
ON o.order_id = oi.order_id
GROUP BY c.customer_id, c.customer_name
ORDER BY revenue DESC
LIMIT 5;
```

```python
%%sql

## Q12 – Customers Who Never Ordered Identify customers who have never placed an order

SELECT
c.customer_id,
c.customer_name
FROM customers c
LEFT JOIN orders o
ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL;
```

```python
%%sql

## Q13 – Above-Average Products Find products whose revenue is above the average revenue across all products. Show product name, product revenue, and average product revenue

WITH product_revenue AS (
SELECT
p.product_id,
p.product_name,
SUM(oi.quantity * oi.unit_price * (1 - oi.discount / 100.0)) AS revenue
FROM products p
JOIN order_items oi
ON p.product_id = oi.product_id
GROUP BY p.product_id, p.product_name
)
SELECT
product_name,
ROUND(revenue, 2) AS product_revenue,
ROUND((SELECT AVG(revenue) FROM product_revenue), 2) AS avg_product_revenue
FROM product_revenue
WHERE revenue > (SELECT AVG(revenue) FROM product_revenue)
ORDER BY revenue DESC;
```

```python
%%sql

 ## Q14 – Monthly Revenue Report Generate a monthly report with year, month, order count, quantity sold, and revenue. Sort chronologically

SELECT
EXTRACT(YEAR FROM o.order_date)  AS order_year,
EXTRACT(MONTH FROM o.order_date) AS order_month,
COUNT(DISTINCT o.order_id)       AS order_count,
SUM(oi.quantity)                 AS quantity_sold,
ROUND(SUM(oi.quantity * oi.unit_price * (1 - oi.discount / 100.0)), 2) AS revenue
FROM orders o
JOIN order_items oi
ON o.order_id = oi.order_id
GROUP BY
EXTRACT(YEAR FROM o.order_date),
EXTRACT(MONTH FROM o.order_date)
ORDER BY order_year, order_month;
```

```python
%%sql

## Q15 – Rank Products by Revenue Rank products by revenue within each category. Show category, product, revenue, and rank. Use RANK or DENSE_RANK

WITH product_revenue AS (
SELECT
p.category,
p.product_name,
SUM(oi.quantity * oi.unit_price * (1 - oi.discount / 100.0)) AS revenue
FROM products p
JOIN order_items oi
ON p.product_id = oi.product_id
GROUP BY p.category, p.product_id, p.product_name
)
SELECT
category,
product_name AS product,
ROUND(revenue, 2) AS revenue,
RANK() OVER (
PARTITION BY category
ORDER BY revenue DESC
) AS revenue_rank
FROM product_revenue
ORDER BY category, revenue_rank;
```
