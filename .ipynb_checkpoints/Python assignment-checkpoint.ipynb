```python
# problem 1: Without using build in functions
# The largest number
# The smallest number
# The total of all numbers
# The average of the numbers

numbers = [45, 12, 78, 34, 89, 23, 56]

# The largest number

largest= numbers[0]
for i in numbers:
 if i>largest:
   largest = i
print(f"Largest number:",largest)

# The smallest number

smallest = numbers[0]
for i in numbers:
  if i<smallest:
    smallest =i
print(f"Smallest number:",smallest)    

# The total of all numbers

numbers = [45, 12, 78, 34, 89, 23, 56]
total = 0
for i in numbers:
 total+=i
print(f"Total:",total)


# The average of the numbers

numbers = [45, 12, 78, 34, 89, 23, 56]
total = 0
for i in numbers:
 total+=i
avg = total/len(numbers)
print(f"Average:",avg)
```

```python
# problem 1: With using build in functions
# The largest number
# The smallest number
# The total of all numbers
# The average of the numbers

numbers = [45, 12, 78, 34, 89, 23, 56]

# The largest number
print(f"Largest number:",max(numbers))

# The smallest number
print(f"Smallest number:",min(numbers))

# The total of all numbers
print(f"Total:",sum(numbers))

# The average of the numbers
total = sum(numbers)
average = total/len(numbers)
print(f"Average:",average) 
```

```python
# Question 2:
# Create a new list containing only unique values 

numbers = [10, 20, 10, 30, 20, 40, 50, 30]
unique_values = []
for i in numbers:
  if i not in unique_values:
    unique_values.append(i)
print(f"Unique values:",unique_values)   

```

```python
# Question 3:
# Find how many times each number occurs

numbers = [10, 20, 10, 30, 20, 10, 40, 20]
unique_values = []
for i in numbers:
  if i not in unique_values:
   unique_values.append(i)
for i in unique_values:
 print(i,"->", numbers.count(i))
```

```python
# Question 4
# Create two lists:
# Even numbers
# Odd numbers


numbers = [10, 15, 22, 33, 40, 51, 64, 75]
Even= []
odd = []
for i in numbers:
 if i%2==0:
  Even.append(i)
 elif i%2 ==1:
  odd.append(i)

print(f"Even:",Even)
print(f"Odd:",odd)
```

```python
# Question 5:
# Find the elements that exist in both lists

list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]

for i in list1:
  for j in list2:
    if i==j:
      print(f"Common elements:",i)
```

```python
# Problem 6:
# Reverse the list without using

numbers = [10, 20, 30, 40, 50]

reverse = []

for i in range(len(numbers)-1,-1,-1):
 reverse.append(numbers[i])
print(f"Reversed List:",reverse) 
```

```python
# Question 8:
# Print every customer's name
# Calculate the total salary
# Find the customer with the highest salary
# Find the average salary.


customers = [
    [1001, "Arun", 25000],
    [1002, "Rahul", 35000],
    [1003, "Priya", 45000],
    [1004, "Vijay", 30000]
]

# Print every customer's name

for i in customers:
  print (f"Every customer name:",i[1])

# Calculate the total salary

total_salary = 0
for i in customers:
  total_salary+=i[2]
print(f"Total salary:",total_salary)

# Find the customer with the highest salary

highest = customers[0][2]
for i in customers:
  if i[2]>highest:
    highest = i[2]
print(f"highest salary:",highest)

# Find the average salary

average = total_salary/len(customers)
print(f"Average salary:",average)
```

```python
# Question 9:
# 1.Print every value
#2.Calculate the total of all values
#3.Find the largest value

matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

# Print every value

for i in matrix:
  for j in i:
    print(f"Every value:",j)

# Calculate the total of all values

total = 0
for i in matrix:
  for j in i:
    total+=j
print(f"Total:",total)

 # Find the largest value

largest = matrix[0][0]
for i in matrix:
  for j in i:
   if j>largest:
    largest =j
print(f"Largest:",largest)  
```

```python
# Question 10: 
# 1.Print each city and its sales
# 2.Calculate total sales
# 3.Find the city with the highest sales


sales = [
    ("Chennai", 10000),
    ("Mumbai", 15000),
    ("Delhi", 12000),
    ("Bangalore", 18000)
]

# Print each city and its sales

for city,amount in sales:
  print(f"Each city and sales:",city,":",amount)

# Calculate total sales

total = 0
for city,amount in sales:
 total+=amount
print(f"Total sales:",total)

# Find the city with the highest sales

top_city,top_sales = sales[0]
for city,amount in sales:
  if amount>top_sales:
    top_city,top_sales = city,amount
print(f"Top city:",top_city) 
print(f"Top sales:",top_sales)   

```
