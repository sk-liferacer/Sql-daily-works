```python
class Animals:
  def __init__(self,name,age):
    self.name = name
    self.age = age
  def intro(self):
    print(f"Hi my name is {self.name} and i am {self.age} years old")
cat = Animals(name = "tom", age =2)
cat.intro()  
```
