```python
%%sql
select 
first_name,last_name,
UPPER(first_name) as Upper_name, 
UPPER(last_name) as Upper_name2  
from employees_new2;
```

```python
%%sql
select 
first_name,last_name,
UPPER(first_name) as Upper_name, 
Length(last_name) as length  
from employees_new2;
```

```python
%%sql
select 
first_name
from employees_new2
where length(first_name)>5 and length(last_name)>5

```

```python
%%sql

select first_name, last_name,
concat (first_name, " ", last_name," -------", email) as name
from employees_new2
```

```python
%%sql
select 
concat_ws("------", first_name, last_name, email, phone) as name
from employees_new2
```

```python
%%sql
select substr(first_name, 2,4) from employees_new2;
```

```python
%%sql
select right(first_name,2
        ) first_name from employees_new2;
```

```python
%%sql
select Rtrim("        hi       ") as hi
```

```python
%%sql
select replace(first_name, 'Arun','Arun kumar') as updated_name
from employees_new2;
```

```python
%%sql
select reverse (first_name), first_name from employees_new2; 
```

```python
%%sql
select locate('a', first_name, 3)as position, first_name 
from employees_new2;
```

```python
%%sql
select concat(first_name, repeat('x',10), last_name) as contacts
from employees_new2
```

```python
%%sql
select concat(first_name, space(20), last_name, space(20), repeat('x', 10)) as name
from employees_new2
```

```python
%%sql
select lpad(employee_id,4,0) as employee_id from employees_new2; 
```
