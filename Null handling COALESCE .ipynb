```python
%%sql
create table students_1(
sid int primary key,
name varchar(50),
phone_number varchar(12),
alternate_number varchar(12),
parent_number varchar(12),
email_id varchar(100),
emg_number varchar(100)
);
```

```python
%%sql
show tables;
```

```python
%%sql
insert into students_1 (sid,name) values (6,'Hema');
insert into students_1 (sid,name,emg_number) values (5,'Priya','9998874563');
insert into students_1 (sid,name,emg_number) values (4,'Divya','123456790');
insert into students_1 (sid,name,phone_number) values (3,'Rajesh','919710051072');
insert into students_1 (sid,name,parent_number) values (2,'Ramesh','91987654321');
insert into students_1(sid,name,email_id) values (1,'Gopal','gopal@gmailcom');
```

```python
%%sql
select* from students_1;
```

```python
%%sql
select COALESCE(phone_number,alternate_number,	parent_number,email_id,emg_number, 'contacts not available') as contacts
from students_1;
```
