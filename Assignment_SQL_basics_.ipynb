```python
%%sql
select employee_name,department,
ifnull(manager_id,0) as manager_id
from employees;
```

```python
Level - 3
```

```python
%%sql
select employee_name,
concat(employee_name, '-', department) as employee_info
from employees; 
```

```python
%%sql
select employee_name,
upper(employee_name) as modified_uppername,
lower(employee_name) as modified_lowername 
from employees;
```

```python
%%sql
select employee_name,
length(employee_name)as length
from employees;
```

```python
%%sql
select employee_name,
substring(employee_name,1,3) as short_name
from employees;
```

```python
%%sql
select employee_name,
replace(employee_name,'a','@') as replaced_name
from employees;
```

```python
%%sql
select employee_name,
trim(employee_name) as modeified_name
from employees
```

```python
Level-4
```

```python
%%sql
select count(*) as employees_count 
from employees;
```

```python
%%sql
select
avg(salary) as avg_salary from
employees;
```

```python
%%sql
select min(salary) as min_salary,max(salary) as max_salary
from employees;
```

```python
%%sql
select sum(salary) as total_salary 
from employees;
```

```python
Level-5
```

```python
%%sql
select count(*) as head_count, department
from employees
group by department;
```

```python
%%sql
select avg(salary) as avg_salary, department
from employees
group by department;
```

```python
%%sql
select min(salary) as min_salary, max(salary) as max_salary, department
from employees
group by department;
```

```python
%%sql
 select sum(salary), department
 from employees
 group by department
 order by sum(salary) desc; 
```

```python
%%sql
 select avg(salary), department
 from employees
 group by department
 having avg(salary)>60000
```

```python
%%sql
 select count(*),department
 from employees
 group by department
 having count(*)>35 
```
